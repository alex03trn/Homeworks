```python
base_url = "https://www.ebi.ac.uk/proteins/api/features"

params1 = {'accession': 'P05067'}
response1 = requests.get(base_url, params=params1)

params2 = {'accession': 'P10636'}
response2 = requests.get(base_url, params=params2)

if response1.status_code == 200 and response2.status_code == 200:
    data1 = response1.json()
    data2 = response2.json()

    seq1 = data1[0]['sequence']
    seq2 = data2[0]['sequence']

    len1 = len(seq1)
    len2 = len(seq2)

    print(f"Белок 1: {data1[0]['entryName']} ({data1[0]['accession']})")
    print(f"Длина: {len1} аминокислот\n")

    print(f"Белок 2: {data2[0]['entryName']} ({data2[0]['accession']})")
    print(f"Длина: {len2} аминокислот\n")

    if len1 > len2:
        diff = len1 - len2
        print(f"Белок APP длиннее Tau на {diff} аминокислот.")
    elif len2 > len1:
        diff = len2 - len1
        print(f"Белок Tau длиннее APP на {diff} аминокислот.")
    else:
        print("Белки одинаковой длины.")

else:
    print("Ошибка при получении данных!")
    print(f"APP: {response1.status_code}, Tau: {response2.status_code}")
```

```python
def get_protein_name(accession):
    
    base_url = "https://www.ebi.ac.uk/proteins/api/features"
    params = {'accession': accession}
    response = requests.get(base_url, params=params)

    if response.status_code == 200:
        data = response.json() 
        return data[0].get('entryName')
    else:
        return f"Ошибка запроса! Белок не найден. Код: {response.status_code}"

accession_id = "P098"
protein_name = get_protein_name(accession_id)
print(f"{accession_id}: {protein_name}")
```

```python
with open("accessions.txt") as f:
    accessions = [line.strip() for line in f if line.strip()]

with open("protein_info.txt", "w", encoding="utf-8") as out:
    for acc in accessions:
        url = f"https://www.ebi.ac.uk/proteins/api/proteins/{acc}"
        r = requests.get(url)
        if r.status_code == 200:
            data = r.json()
            name = data.get('id')
            organism = next((n['value'] for n in data.get('organism', {}).get('names', []) if n['type']=='scientific'))
            out.write(f"{acc}\t{name}\t{organism}\n")
        else:
            out.write(f"{acc}\tОшибка {r.status_code}\n")
        time.sleep(0.5)
```

```python
import xml.etree.ElementTree as ET
fetch_url = "https://eutils.ncbi.nlm.nih.gov/entrez/eutils/efetch.fcgi"

for article_id in id_list:
    fetch_params = {'db': 'pubmed', 'id': article_id, 'retmode': 'xml'} #efetch.fcgi для PubMed не поддерживает json
    response = requests.get(fetch_url, params=fetch_params)
    
    if response.status_code == 200:
        root = ET.fromstring(response.text)
        # Заголовок
        title_elem = root.find(".//ArticleTitle")
        title = title_elem.text if title_elem is not None else "Нет заголовка"
        # Авторы
        authors = []
        for author in root.findall(".//Author"):
            last = author.find("LastName")
            first = author.find("ForeName")
            if last is not None and first is not None:
                authors.append(f"{first.text} {last.text}")
        print(f"\nArticle ID {article_id}")
        print(f"Title: {title}")
        print(f"Authors: {', '.join(authors)}")
    else:
        print(f"Ошибка запроса для ID {article_id}, код {response.status_code}")
    
    time.sleep(0.5)
```

```python
import csv
gene_name = "APOE"

requestURL = f"https://www.ebi.ac.uk/proteins/api/coordinates?offset=0&size=100&gene={gene_name}"

# Отправляем запрос
r = requests.get(requestURL, headers={"Accept": "application/json"})

if r.status_code == 200:
    data = r.json()

    # Извлекаем нужную информацию
    rows = []
    for protein in data:
        accession = protein.get("accession")
        entry_name = protein.get("entryName")
        protein_name = protein.get("protein", {}).get("recommendedName", {}).get("fullName", {}).get("value", "")
        link = f"https://www.uniprot.org/uniprot/{accession}"
        rows.append([accession, entry_name, protein_name, link])

    # Сохраняем в CSV
    csv_filename = f"{gene_name}_proteins.csv"
    with open(csv_filename, "w", newline="", encoding="utf-8") as f:
        writer = csv.writer(f)
        writer.writerow(["UniProt Accession", "Entry Name", "Protein Name", "UniProt Link"])
        writer.writerows(rows)

    print(f"✅ Результат сохранён в {csv_filename}")
else:
    print(f"Ошибка! Код: {r.status_code}, текст: {r.text}")
```

```python
import csv

gene = "APOE"
params = {
    "gene": gene,
    "taxid": 9606,
}
url = "https://www.ebi.ac.uk/proteins/api/coordinates"

response = requests.get(url, headers={"Accept": "application/json"}, params=params)

if response.status_code != 200:
    print("Ошибка запроса:", response.status_code)
    exit()

data = response.json()
rows = []

for item in data:
    accession = item.get("accession")
    name = item.get("name") 
    link = "https://www.uniprot.org/uniprot/{accession}"
    rows.append([accession, name, link])

# Сохранение
with open(f"{gene}_proteins.csv", "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerow(["Accession", "Name", "UniProt Link"])
    writer.writerows(rows)

print("Результат сохранён в {gene}_proteins.csv")
```

```python
import requests

gene_name = input("Введите название гена: ").strip()

url = "https://www.ebi.ac.uk/proteins/api/proteins"
params = {
    "gene": gene_name,
    "taxid": 9606,  
    "size": 10
}
headers = {"Accept": "application/json"}

response = requests.get(url, headers=headers, params=params)
if response.status_code != 200:
    print("Ошибка запроса к EBI Proteins API:", response.status_code)
    exit()

proteins = response.json()
if not proteins:
    print(f"Белки для гена '{gene_name}' не найдены.")
    exit()

first_protein = proteins[0]
accession = first_protein.get("accession", "Неизвестно")

protein_names = first_protein.get("protein", {}).get("submittedName", [])
if protein_names and "fullName" in protein_names[0]:
    protein_name = protein_names[0]["fullName"].get("value", "Неизвестно")
else:
    protein_name = "Неизвестно"

print(f"Первый белок для гена '{gene_name}': {protein_name} (Accession: {accession})")

pubmed_url = "https://eutils.ncbi.nlm.nih.gov/entrez/eutils/esearch.fcgi"
params_pubmed = {
    "db": "pubmed",
    "term": f"{gene_name}[Title/Abstract]", 
    "retmax": 20,
    "retmode": "json"
}

r_pub = requests.get(pubmed_url, params=params_pubmed)
if r_pub.status_code != 200:
    print(f"Ошибка при запросе к PubMed: {r_pub.status_code}")
    exit()

pubmed_data = r_pub.json()
pmids = pubmed_data.get("esearchresult", {}).get("idlist", [])

if pmids:
    print("Найденные PubMed ID статей:")
    for pmid in pmids:
        print(pmid)
else:
    print("Статьи не найдены.")
```

```python
esearch_url = "https://eutils.ncbi.nlm.nih.gov/entrez/eutils/esearch.fcgi"
params_es = {
    "db": "pubmed",
    "term": '"PARP1"[Title] AND "cardiovascular diseases"[Title/Abstract]',
    "retmax": 10,
    "sort": "pub date",
    "retmode": "json"
}
r_es = requests.get(esearch_url, params=params_es)
pmids = r_es.json().get("esearchresult", {}).get("idlist", [])

esummary_url = "https://eutils.ncbi.nlm.nih.gov/entrez/eutils/esummary.fcgi"
params_sum = {"db": "pubmed", "id": ",".join(pmids), "retmode": "json"}
r_sum = requests.get(esummary_url, params=params_sum)
sum_data = r_sum.json().get("result", {})

articles_found = 0
print("Последние статьи с PARP1")
for pmid in pmids:
    doc = sum_data.get(pmid, {})
    title = doc.get("title", "")
    if "PARP1" in title:
        print(f"{pmid}: {title}")
        articles_found += 1
        if articles_found >= 3:
            break

if articles_found == 0:
    print("Не найдено статей с PARP1 в заголовке.")
```
